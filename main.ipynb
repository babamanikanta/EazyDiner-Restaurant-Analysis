# 6. Cost comparison across cities

This box plot compares restaurant costs across the top five cities by restaurant count. It helps identify differences in typical dining costs, price variations, and outliers.

# 9. Top 10 cities by median cost

This chart compares the median dining cost across the 10 cities with the highest median costs. It helps identify cities where typical restaurant prices are relatively higher.

```python
median_cost = (
    df.groupby("Search_city")["Average_cost_for_two"]
    .median()
    .sort_values(ascending=False)
    .head(10)
)

plt.plot(median_cost.index, median_cost.values, marker="o")

plt.title("Top 10 Cities by Median Restaurant Cost")
plt.xlabel("City")
plt.ylabel("Median Cost for Two")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# 10. Top 10 areas by restaurant count

This horizontal bar chart displays the top 10 areas with the highest number of restaurant listings, along with their cities. It helps identify areas with a greater concentration of listed restaurants.



```python
area_counts = (
    df.groupby(["Area", "City"])
      .size()
      .reset_index(name="Restaurant_Count")
      .nlargest(10, "Restaurant_Count")
      .sort_values("Restaurant_Count")
)

labels = area_counts["Area"] + "\n"+" (" + area_counts["City"] + ")"

plt.figure(figsize=(10, 6))
plt.barh(labels, area_counts["Restaurant_Count"])

plt.title("Top 10 Areas by Restaurant Count")
plt.xlabel("Restaurant Count")
plt.ylabel("Area (City)")
plt.tight_layout()
plt.show()
```

# 12.Average restaurant rating by city

This bar chart compares the average restaurant ratings across cities. It helps identify cities with higher average ratings among restaurants with available rating data.

```python
# 1. Group by city and area
area_rating = (
    df.groupby(["Search_city", "Area"])
    .agg(
        Average_Rating=("Rating", "mean"),
        Restaurant_Count=("Rating", "count")
    )
    .reset_index()
)

# 2. Keep areas having at least 5 rated restaurants
area_rating = area_rating[
    area_rating["Restaurant_Count"] >= 5
]

# 3. Get top 10 areas
top_areas = area_rating.nlargest(10, "Average_Rating")

# 4. Display results
print(top_areas)

# 5. Plot
labels = top_areas["Area"] + " (" + top_areas["Search_city"] + ")"

plt.figure(figsize=(12, 6))

bars = plt.barh(
    labels,
    top_areas["Average_Rating"]
)

plt.bar_label(bars, fmt="%.2f", padding=3)

plt.xlabel("Average Rating")
plt.ylabel("Area")
plt.title("Top 10 Areas by Average Restaurant Rating")
plt.xlim(0, 5.5)
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

# 13. Top 10 Cuisines by Average Restaurant Cost
This horizontal bar chart compares the average dining cost for different cuisines. It helps identify cuisines associated with higher restaurant prices.

```python
# Calculate average cost and restaurant count for each cuisine
cuisine_cost = (
    df.explode("Cuisines")
    .groupby("Cuisines")["Average_cost_for_two"]
    .agg(["mean", "count"])
)

# Keep cuisines with at least 10 restaurants
cuisine_cost = cuisine_cost[cuisine_cost["count"] >= 10]

# Select the top 10 cuisines by average cost
top_cuisines = (
    cuisine_cost.sort_values("mean", ascending=False)
    .head(10)
    .sort_values("mean", ascending=True)
)

print(top_cuisines.round(2))

# Plot
plt.figure(figsize=(11, 6))

bars = plt.barh(
    top_cuisines.index,
    top_cuisines["mean"],
    color="steelblue"
)

# Display average cost on each bar
plt.bar_label(bars, fmt="₹%.0f", padding=4)

plt.title("Top 10 Cuisines by Average Restaurant Cost")
plt.xlabel("Average Cost for Two (₹)")
plt.ylabel("Cuisine")

plt.margins(x=0.15)
plt.tight_layout()
plt.show()
```
